```python
# Implemenasi Has Table Sederhana

class SimpleHashTable:
  def __init__(self, size):
      self.size = size # menyimpan ukuran hash table
      self.table = [None] * size # membuat list kosong sesuai ukuran tale

  def hash_function(self, key):  # mengubah key menjadi nomor inde supaya data bisa disimpan diposisi mana
      return hash(key) % self.size

  def insert(self, key, value):
      index = self.hash_function(key) # mencari posisi atau index untuk data
      self.table[index] = value # menyimpan value ke dalam data

  def search(self, key):
      index = self.hash_function(key) # mencari posisi data berdasarkan key
      return self.table[index] # mengambil data dari posisi tersebut

# Contoh penggunaan
ht = SimpleHashTable(10) # membuat hash table degan ukuran 10
ht.insert("apel", 100) # menyimpan data apel yang memiliki nilai 100
ht.insert("pisang", 200)

print("Nilai untuk 'apel':", ht.search("apel")) # mencari nilai dari key "apel"
print("Nilai untuk 'pisang':", ht.search("pisang"))
```

```python
# Penanganan Colluision dengan Linear Probing

class LinearProbingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        original_index = index # menyimpan index awal dan digunakan untuk mengecek apakah table sudah penuh

        while self.table[index] is not None: # selama slot masih terisi maka cari slot lain
          if self.table[index] [0] == key: # jika key sudah ada maka data lama akan diganti
            break
          index = (index + 1) % self.size # pindah ke index berikutnya
          # linear probing = maju satu-satu mencari tempat kosong

          if index == original_index:
              raise Exception("Hash table penuh") # jika kembali ke index awal berarti semua slot penuh

        self.table[index] = (key, value) # menyimpan key dan value ke slot yang ditentukan.

    def search(self, key):
        index = self.hash_function(key) # menari posisi awal berdasar
        original_index = index

        while self.table[index] is not None:
            if self.table[index] [0] == key:
                return self.table[index] [1]
            index = (index + 1) % self.size
            if index == original_index:
                break
        return None

# Contoh penggunaan
ht = LinearProbingHashTable(5)
ht.insert("apel", 100)
ht.insert("pisang", 200)
ht.insert("melon", 300)

print("Nilai untuk 'apel':", ht.search("apel"))
print("Nilai untuk 'melon':", ht.search("melon"))
```

```python
# Penanganna Collision dengan Chaining

class ChainingHashTable:
    def __init__(self, size):
      self.size = size
      self.table = [[] for _ in range(size)]

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        for i, (k, v) in enumerate (self.table[index]): # mengecek isi slot satu persatu
                                # enumerate() untuk mengambil nomor index + isi data sekaligus

            if k == key:
                self.table[index] [i] = (key, value) # jika key sudah ada maka value lama diganti
                return
        self.table[index].append((key, value)) # jika key belum ada maka data ditamabhkan ke slot

    def search(self, key):
        index = self.hash_function(key) # mencari posisi berdasarkan key
        for k, v in self.table[index]:
            if k == key:
                return v
        return None

  # Contoh enggunnaan

ht = ChainingHashTable(5)
ht.insert("apel", 100)
ht.insert("pisang", 200)
ht.insert("melon", 300)

print("Nilai untuk 'apel':", ht.search("apel"))
print("Nilai untuk 'melon':", ht.search("melon"))
```
