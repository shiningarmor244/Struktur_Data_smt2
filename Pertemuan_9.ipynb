```python
# Representasi Graf dengan Adjecency List

# kalau dua node saling terhubung oleh garis/edge, maka kedua node tersebut disebut adjacent (bertetangga).

# Membuat Graf tak berarah menggunakan adjencecy list

graph = {               # membuat dictionary python
    'A': ['B', 'C'],    # 'node': [daftar tetangga]
    'B': ['A', 'D', 'E'], # artinya → node 'B' terhubung ke A, D, dan E
    'C': ['A'],         # node 'C' terhubung ke A
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

# Menampilkan graf
def print_graph(graf): # membuat fungsi yang digunakan untuk menmpilkan isis graf
  for node in graf: # mengambil setiap node satu-persatu yang nanti akan ditampilkan
    print({f"{node} -> {graf[node]}"}) # menampilkan nama node dan daftar tetangganya

    # print("A → ['B', 'C']")
    # saat node = 'E'
    # maka dibaca graf[node] → mengambil daftar tetangga dari node tersebut
    # graf['E']

print_graph(graph)
```

```python
# Traversal Graf menggunakan BFS (Breadth First Search)
# mengunjungi node secara melebar atau per level di mulai dari node awal lalu bergerak ke tetangga terdekat

from collections import deque # Digunakan untuk mengimpor deque
                              # deque digunakan untuk queue konsel FIFO

def bfs(graph, start): # start → node awal tranversal
    visited = set() # membuat set kosong untuk menyimpan node yang telah ditelusuri
    queue = deque ([start]) # queue lalu memasukkan node awal

    while queue:
        vertex = queue.popleft() # mengambil node paling depan dari queue
                                # Queue = [A, B, C]
                                # popleft()
                                # vertex = A
                                # Queue = [B, C]
        if vertex not in visited:
            print (vertex, end=" ")
            visited.add(vertex) # menambahkan node ke dalam visited
            queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])
            # graph[vertex] → graph['A'] → hasilnya ['B', 'C']
            # if neighbor not in visited digunakan supaya node yang sudah pernah dikunjungi tidak dimasukkan lagi
            # queue.extend(...) → digunakan unruk menambahka semua tetangga ke queue


# contoh pemanggilan
bfs(graph, 'A')
```

```python
# Traversal Graf menggunakan DFS (Depth First Search)
# menyelam sedalam mungkin ke satu jalur terlebih dahulu baru kembli untuk mengecek jalur lainnya

def dfs(graph, start, visited=None): # start → node awal, visited → menyimpan node yang sudah dikunjungi
  if visited is None:
      visited = set ()

  if start not in visited:
      print (start, end=" ") # menampilkan node tranversal. end=" " → supaya output nyambung
      visited.add(start) # menandai node yang sudah dikunjungi
      for neighbor in graph[start]: # mengambil semua tetangga dari node
        dfs(graph, neighbor, visited) #rekursif dfs

# Contoh pemanggilan
dfs(graph, 'A')
```

```python
# 2. Implementasikan DFS

def dfs(graph, start, visited=None): # start → node awal, visited → menyimpan node yang sudah dikunjungi
  if visited is None:
      visited = set ()

  if start not in visited:
      print (start, end=" ") # menampilkan node tranversal. end=" " → supaya output nyambung
      visited.add(start) # menandai node yang sudah dikunjungi
      for neighbor in graph[start]: # mengambil semua tetangga dari node
        dfs(graph, neighbor, visited) # rekursif dfs

# Contoh pemanggilan
dfs(graf_1, 'APEL')
```

```python
# 3. Modifikasi fungsi BFS agar mengembalikan list urutan kunjungan, bukan hanya mencetak.
# BFS biasa hanya mencetak urutan traversal langsung ke layar, sedangkan BFS yang dimodifikasi
# menyimpan hasil traversal ke dalam list sehingga data traversal dapat digunakan kembali untuk proses lain.

from collections import deque

def bfs_list(graph, start):

    visited = set() # menyimpan node yang sudah dikunjungi
    queue = deque([start]) # queue untuk BFS
    result = [] # menyimpan hasil urutan traversal

    while queue:

        vertex = queue.popleft() # mengambil node paling depa dari queue
        if vertex not in visited:

            visited.add(vertex)

            result.append(vertex)
            # memasukkan node ke dalam list hasil traversal

            queue.extend([
                neighbor for neighbor in graph[vertex]
                if neighbor not in visited
            ]) # digunakan untuk memasukkan semua tetangga node ke queue

    return result

# Menampilkan hasil BFS dalam bentuk list
hasil = bfs_list(graf_1, 'APEL')

print("Urutan BFS:")
print(hasil)
```

```python
# 4. Buat fungsi find_path(graph, start, end) yang mengembalikkan satu jalur
# dari start ke end jika ada menggunakan DFS

def find_path(graph, start, end, path=None): # end → node tujuan, path → menyimpan jalur tranversal sementara

    if path is None:
        path = []

    path.append(start)
    if start == end: # jika node sekarang sama dengan tujuang maka jalur langsung dikembalikan
        return path

    if start not in graph: # pencarian diberhentikan
        return None

    for neighbor in graph[start]: # mengambil semua tetangga dari node sekarang

        if neighbor not in path: # menghindari loop tak terbatas

            new_path = find_path(
                graph,
                neighbor, # fungsi rekursif
                end,
                path.copy()
            )

            if new_path:
                return new_path

    return None


hasil = find_path(graf_1, 'APEL', 'NANGKA') # mencari jalur dari APEL → NANGKA

print("Jalur dari APEL ke NANGKA:")
print(hasil)
```

```python
# DFS untuk mengecek koneksi graf

def dfs(graph, start, visited):

    visited.add(start)

    for neighbor in graph[start]:

        if neighbor not in visited:
            dfs(graph, neighbor, visited)


def is_connected(graph): # fungsi utama untuk mengecek apakah graf connected atau tidak

    visited = set()

    start_node = next(iter(graph)) # mengambilnode awal secara ototmatis
                    # iter →iterator, membuat iterator dari garph

    dfs(graph, start_node, visited)

    return len(visited) == len(graph)
    # le(visited) → jumlah node yang berhasil dikunjungi
    # len(garph) → jumlah seluruh node dalam graf
    # jika jumlahnya SAMA → True → artinya semua node saling terhubung
    # Jika jumlanya berbeda → False

print(is_connected(graf_1))
```
