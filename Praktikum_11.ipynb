```python
# Case : Coin Change

# Metode penyelesaian masalah yang selalu memilih solusi terbaik pada saat itu juga (local optimum)
# tanpa mempertimbangkan kemungkinan pilihan lain di masa depan.

def coin_change_greedy(amount, coins): # membuat fungsi coin_change_greedy
                        # amount → jumlah uang yang ingin ditukar
                        # daftar pecahan koin yang tersedia
    coins.sort (reverse = True) # mengurutkan koin dari yang terbesar ke terkecil (reverse = True → urutan menurun)
    result = []
    for coin in coins:
        while amount >= coin: # selama nilai uang masih cukup untuk menggunakan koin tersebut
            amount -= coin # mengurangi jumlah yang tersisa ex: 57 - 25 = 32
            result.append (coin) # menyimpan koin yang terpilih
    return result

# Contoh Penggunaan.
amount = 57 # menentukan jumlah uang yang ingin keluar
coins = [25, 10, 5, 1] # menentukan pecahan koin yang tersedia
change = coin_change_greedy(amount, coins) # menjalankan fungi greedy
print("Koin yang digunakan: ", change) # menampilkan hasil koin
```

```python
# Factional Knapsack

def fractional_knapsack(items, capacity): # items → daftar barang, capacity → kasitas maksimum tas
    # itemns = list of tuples (value, weight) # setiap barang disimpan dalam bentuk value, weight
    items = sorted(items, key = lambda x: x[0]/x[1], reverse = True) # mengurutkan barang berdasakran rasio value
    # rumus: nilai ÷ berat
    # semakin besar hasilnya semakin diprioristaskan
    total_value = 0.0
    for value, weight in items: # weight → berat barang
        if capacity >= weight: # mengecek apakah kapasitas tas masih cukup
            capacity -= weight # kapasitas tas dikurangi berat barang
            total_value += value # menambahkan seluruh nilai barang
        else:
            total_value += value * (capacity / weight) # mengambil sebagian besar barang
            # rumus:
            # nilai × persentase bagian yang diambil
    # contoh:
            # nilai = 120
            # berat = 30
            # kapasitas tersisa = 20

            # maka:
            # 120 × (20/30)
            break
    return total_value

# Contoh penggunaan.
items = [(60, 10), (100, 20), (120, 30)] # (nilai, berat) daftar barang
capacity = 50 # kapasitas maksimum tas
max_value = fractional_knapsack(items, capacity) # menjalankan algoritma greedy
print("Nilai maksimum yang dapat dibwa:", max_value)
```

```python
# Minimum Spanning Tree: Prim dan Kruskal
# Pohon yang menghubungkan seluruh node dalam graf dengan total bobot minimum tanpa membentuk siklus

# Contoh Kode Program 1

import heapq # library heapq

def prim_mst(graph, start) : # pada Prim edge dengan bobot terkecil selalu dipilih lebih dulu
    visited = set ()
    min_heap = [(0, start)] # membuat priority queue awal (bobot, node)
    total_weight = 0

    while min_heap:
        weight, node = heapq.heappop(min_heap) # mengambil data dengan bobot terkecil
        if node in visited:
            continue
        visited.add(node) # menandai node sudah dikunjungi
        total_weight += weight # bobot edge ke MTS
        for neighbor, edge_weight in graph[node]:
            if neighbor not in visited:
                heapq.heappush(min_heap, (edge_weight, neighbor))
    return total_weight

# Representasi graf: adjacency list
graph = {
    'A': [('B', 2), ('C', 3)],
    # A terhubung ke B dengan bobot 2
    'B': [('A', 2), ('C', 1), ('D', 1)],
    'C': [('A', 3), ('B', 1), ('D', 4)],
    'D': [('B', 1), ('C', 4)]
}

print("Total bobot MST (Prim):", prim_mst(graph, 'A'))
```

```python
# Contoh Kode Program 2

def kruskal_mst(edges, n_nodes) :
  # edges → daftar edge pada graf
  # n_nodes → jumlah node dalam graf

    parent = {i: i for i in range (n_nodes)} #  membuat struktur parent

    def find(x): # mencari akar (root)
      while parent[x] != x: # selama parent bukan dirinya sendiri
          x = parent[x] # bergerak menuju parent berikutnya
      return x # mengembalikan root

    def union(x, y) : # fungsi untuk menggabunykan 2 kelompk
      root_x = find(x) # mencari root x
      root_y = find(y)
      if root_x != root_y: # jika root berbeda
        # artinya kedua node berada
        # pada kelompok yang berbeda

          parent[root_y] = root_x # menggabungkn kelompok
           # root_y menjadi anak root_x
          return True
      return False

    edges.sort(key = lambda x: x[2]) # mengurutkan edge berdasarkan bobit.
    # x[2] = weight

    total_weight = 0
    for u, v, weight in edges: # mengambil setiap edge satu per satu
        if union(u, v):
            total_weight += weight # tambahkan bobot ke mst
    return total_weight

# edges: (node1, node2, weight), nodes are represeted by integers
edges = [(0, 1, 2), (0, 2, 3), (1, 2, 1), (1, 3, 1), (2, 3, 4)]
print("Total bobot MST (Kruskal): ", kruskal_mst(edges, 4))
```
