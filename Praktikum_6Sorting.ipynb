```python
# Implememntasi Buble sort

def bubble_sort(arr): # Membuat fungsi baru bernama bubble_sort
  n = len(arr) # variabel yang akan menyimpan jumlah elemen
  for i in range(n - 1): # untuk mengulang proses pengurutan (loop luar)
    for j in range (n - i - 1): # untuk membandingkn elemen satu persatu (loop dalam)
    # setiap putaran elemen terbesah sudah "pindah ke belakang"

      if arr[j] > arr[j + 1]: # Membandingkan 2 elemen yang bersebelahan kalau kiri lebih besar berarti urutannya salah
        arr[j], arr[j + 1] = arr[j + 1], arr[j] # ini untuk switching misal 5 3 → 3 5

  return arr # mengembalikkan array yang sudah diurutkan

# COntoh penggunaan

data = [64, 34, 25,12, 22, 11, 90] # ini list
sorted_data = bubble_sort(data.copy()) # Menggunakan copy agar data aslu tidaj berubah
print("Data sebelum sorting:", data)
print("Data setelah Bubble Sort:", sorted_data)
```

```python
# Implementasi Selection Sort

def selection_sort(arr): # Membuat fungsi baru baru bernama sorting_sort
  n = len(arr)
  for i in range(n - 1):
    min_index = i # ini untuk posisi sementara dimana posisi i adalah yang paling kecil
    for j in range(i +1, n): # mencari nilai yang lebih kecil dari i mulai dari kanan
      if arr[j] < arr[min_index]: # untuk mengecek nilai lebih kecil
        min_index = j # indeks untuk menyimpan nilai lebih kecil

    arr[i], arr[min_index] = arr[min_index], arr[i] # nilai terkecil akan ke depan, lebih besar ke belakang (tukar)

  return arr
# Contoh penggunaan
data = [64, 25, 12, 22, 11]
sorted_data = selection_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data seteleh Selection Sort:", sorted_data)

# Dia mencari minimum
# penukaran hanya 1x per putaran
# lebih sedikit swap
```

```python
# Implementation Insertion Sort
def insertion_sort(arr): # Membuat fungsi baru bernama insertion_sort
  n = len(arr)
  for i in range(1, n): # loop mulai dari indeks -1 (minus 1) karena indeks 0 itu dianggap sudah urut
    key = arr[i]
    j = i - 1 # menunjuk elemn selanjutnya dan akan dibandingkan dengan key
    while j >= 0 and arr[j] > key: # selama masih di dalam array dan nilai besar maka akan bergeser terus
      arr[j + 1] = arr[j] # menggeser nilai ke kanan
      # [34, 25] → jadi → [34, 34]

      j -= 1 # geser ke kiri
    arr[j + 1] = key # menaruh key ke posisi yang benar, tempat kosong digeser-geser maka ada slot kosong
  return arr

  # Contoh penggunaan
data = [64, 34, 25, 12, 22, 11, 90]
sorted_data = insertion_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data setelah Insertion Sort:", sorted_data)
```

```python
# Implementasi Merge sort
def merge_sort(arr): # Membuat fungsi baru
  if len(arr) > 1: # kalau data lebih dari 1 harus diurutkan tapi kalau cuma 1 berarti sudah urut
      mid = len(arr) // 2 # rumus mencari titik tengah
      left_half = arr[:mid] # array kiri
      right_half = arr[mid:] # array kanan (karena arraynya dibagi menjadi 2 bagian)

      merge_sort(left_half) # ini rekursif. fungsi memanggil dirinya sendiri
      merge_sort(right_half)

      i = j = k = 0 # i → kiri, j → kanan, k → array utama

      while i < len(left_half) and j < len(right_half):
        if left_half[i] < right_half[j]: # membandingkan elemen kiri dan kanan
            arr[k] = left_half[i] # mengambil yang lebih kecil dari sisi kiri
            i +=1
        else:
            arr[k] = right_half[j] # mengambil yang lebih kecil dari kanan
            j += 1
        k += 1 # maju ke posisi berikutnya di array

      while i < len(left_half):  # kalau masih ada sisa di kiri
        arr[k] = left_half[i] # masukkan ke array utama
        i += 1 #
        k += 1

      while j < len(right_half):
        arr[k]= right_half[j] # masukkan ke array utama
        j += 1
        k += 1

  return arr

  # Contoh penggunaan
data = [64, 25, 12, 22, 11]
sorted_data = merge_sort(data.copy())
print("Data sebelum sorting: ", data)
print("Data setelah Merge Sort: ", sorted_data)
```

```python
# Implementasi Quick Sort

def quick_sort(arr):
    if len(arr) <= 1: # menghentikan rekursif
        return arr

    pivot = arr[len(arr) // 2] # untuk memilij nolai pivot, pivot itu nilai tengah array
    left = [x for x in arr if x < pivot] # nilai lebih kecil
    middle = [x for x in arr if x == pivot] # nilai yang sama dengan pivot
    right = [x for x in arr if x > pivot] # nilai lebih besar

    return quick_sort (left) + middle + quick_sort(right) # mengurutkan bagian kiri, bagian kanan dan digabung

# contoh penggunaan

data = [64, 25, 12, 22, 11]
sorted_data = quick_sort(data.copy())
print("Data sebelum sorting: ", data)
print("Data setelah quick Sort: ", sorted_data)
```
