```python
# Binary Tree secara manual

class Node: # membuat class baru bernama Node
  def __init__(self, data) :
    self.data = data # menyimpan data
    self.left = None # digunakan untuk menghubungkan node ke cabang kiri (left). None digunakan dengan arti bahwa cabang disebelah kiri masih kosong
    self.right = None # cabang kanan (right)

root = Node(1) # membuat root/node dengan nilai 1
root.left = Node(2) # anak (pertama) kiri dari root bernuilai 2
root.right = Node(3) # anak (pertama) kanan dari root bernilai 3
root.left.left = Node(4) # anak (pertama) dari node 2 bernilai 4
root.left.right = Node(5) # anak (pertama) dari node 2 bernilai 5

# Fungsi inoder traversal
def inorder(node) : # membut fungsi inorder traversal
  if node : # Kalau node ada/is not None maka lanjut.
    inorder (node.left) # mengunjungi subtree kiri lalu dicek
    print(node.data, end=' ')
    inorder(node.right)

print("Hasil inoder travesal dari binary tree :")
inorder (root)
```

```python
# Membuat Binary SEARCH TREE (BST)
# Kosepnya yaitu seperti pohon dengan cara nilai kecil akan masuk ke kiri dan nilai besar akan masuk ke kanan.

class Node:
  def __init__(self, data) :
    self.data = data
    self.left = None
    self.right = None

class BinarySearchTree:
  def __init__(self):
    self.root = None # root awal masih kosong
  def insert(self, root, data) : # fungsi untuk memasukkan data ke BST
    if root is None: # ini kalay node kosong
      return Node(data)
    if data < root.data: # kalau data lebih kecil dari root maka masuk ke kiri
      root.left = self.insert(root.left, data) # memanggil fungsi lagi untuk memasukkan data ke subtree kiri
    else: # kalau lebih besar masuk ke kanan
      root.right = self.insert(root.right, data)
    return root

bst = BinarySearchTree() # membuat objek BST
root = None
data_list = [50, 30, 70, 20, 40, 60, 80] # data yang akan dimasukkan ke BST
for value in data_list: # mengambil data satu persatu dari list
    root = bst.insert(root, value)
```

```python
# Travesal pada BST

# traversal : inorder, preorder, dan postorder

# Inorder: kiri → root → kanan
# Preorder: root → kiri → kanan
# Postorder: kiri → kanan → root

def inorder(node) :
  if node:
    inorder (node.left) # mengunjungi subtree kiri dulu
    print(node.data, end=' ') # menampilakn data node. ' ' digunakan supaya otuput menyambung
    inorder (node.right)
    # urutan kiri → root → kanan

def preorder(node):
  if node:
    print(node.data, end=' ') # root dicetak lebih dulu
    preorder (node.left) # lalu lanjut ke kiri
    preorder (node.right) # ke kanan
    # urutan root → kiri → kanan

def postorder (node) :
  if node:
    postorder (node.left)
    postorder (node.right)
    print (node.data, end=' ')
    # urutan iri → kanan → root

print("Inorder Traversal :")
inorder (root)
print("\nPreorder Traversal : ")
preorder (root)
print("\nPostorder Traversal : ")
postorder (root)
```

```python
# PENCARIAN NIlAI DALAM BST
# Mencari nilai tertentu
# kiri lebih kecil, kanan lebih besar

def search(node, key): # node → node sekarang, key → nilai yang dicari
  if node is None or node.data == key:
    return node
  if key < node.data: # kalau nilai lebih kecil dari node sekarang maka cari di kiri
    return search(node.left, key) # memanggil rekursif ke subtree kiri
  return search(node.right, key) # kalau tidak lebih kecil maka cari di kanan

# Uji Pencarian
key = 60
result = search(root, key) # menjalanakn fungsi pencarina mulai dari root.
if result:
  print(f"{key} ditemukan dalam tree.")
else:
  print(f"{key} tidak ditemukan.")

key = 25
result = search(root, key)
if result:
  print(f"{key} ditemukan dalam tree.")
else:
  print(f"{key} tidak ditemukan.")
```

```python
#SOAL 2

# Buat program Python untuk membentuk BST secara otomatis dari data numerik yang unik bagi Anda.
# Ketentuan:
# • Buat daftar 7 angka dengan aturan berikut:
# 1. Tiga digit terakhir NPM Anda
# 2. Tanggal lahir (DDMM, pisahkan menjadi dua angka: DD dan MM)
# 3. ASCII dari huruf pertama dan kedua nama depan Anda
# 4. Tambahkan dua angka bebas (boleh favorit Anda)
# • Sisipkan semua angka tersebut ke dalam BST.
# • Tampilkan hasil traversal:
# o Inorder
# o Preorder
# o Postorder
# • Tambahkan fitur pencarian:
# o Cari angka dengan nilai dua digit terakhir NPM Anda
# o Cari angka yang tidak termasuk dalam list (buat sendiri)

# Output yang Diharapkan:
# • Hasil setiap traversal tercetak rapi
# • Hasil pencarian menunjukkan apakah data ditemukan atau tidak
# • Tambahkan komentar penjelasan pendek di setiap bagian kode


# SOAL 2
# Membentuk Binary Search Tree (BST)

# Membuat node
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


# Membuat class BST
class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, root, data):
        if root is None:
            return Node(data)

        # Jika data lebih kecil masuk ke kiri
        if data < root.data:
            root.left = self.insert(root.left, data)

        # Jika data lebih besar masuk ke kanan
        else:
            root.right = self.insert(root.right, data)

        return root


# Fungsi inorder traversal
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)


# Fungsi preorder traversal
def preorder(node):
    if node:
        print(node.data, end=' ')
        preorder(node.left)
        preorder(node.right)


# Fungsi postorder traversal
def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.data, end=' ')


# Fungsi pencarian pada BST
def search(node, key):

    # Jika node kosong atau data ditemukan
    if node is None or node.data == key:
        return node

    # Jika key lebih kecil cari ke kiri
    if key < node.data:
        return search(node.left, key)

    # Selain itu cari ke kanan
    return search(node.right, key)



# 1. Tiga digit terakhir NPM = 031 → 31
# 2. Tanggal lahir = 21 dan 03
# 3. ASCII huruf pertama dan kedua nama depan
#    I = 73
#    L = 76
# 4. Dua angka bebas = 99 dan 50

data_list = [31, 21, 3, 73, 76, 99, 50]


# Membuat object BST
bst = BinarySearchTree()

# Root awal kosong
root = None


# Memasukkan semua data ke BST
for value in data_list:
    root = bst.insert(root, value)



# TRAVERSAL
print("Inorder Traversal:")
inorder(root)

print("\n\nPreorder Traversal:")
preorder(root)

print("\n\nPostorder Traversal:")
postorder(root)



# PENCARIAN DATA
# Mencari dua digit terakhir NPM
key = 31
result = search(root, key)

if result:
    print(f"\n\n{key} ditemukan dalam BST.")
else:
    print(f"\n\n{key} tidak ditemukan.")


# Mencari angka yang tidak ada dalam list
key = 77
result = search(root, key)

if result:
    print(f"\n{key} ditemukan dalam BST.")
else:
    print(f"\n{key} tidak ditemukan.")




```
